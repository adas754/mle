```python
# how to check the null value
df.isnull().sum()
```

```python
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
df['Married'] = df['Married'].fillna(df['Married'].mode()[0])
```

```python
df['Dependents'] = df['Dependents'].fillna(df['Dependents'].mode()[0])
```

```python
df['Self_Employed'] = df['Self_Employed'].fillna(df['Self_Employed'].mode()[0])
```

```python
df['LoanAmount'].describe()
```

```python
df.loc[df['LoanAmount'].isnull(),'LoanAmount']=df['LoanAmount'].median() 
```

```python
df.loc[df['Loan_Amount_Term'].isnull(),'Loan_Amount_Term']=df['Loan_Amount_Term'].median() 
```

```python
df['Credit_History'].describe()
```

```python
df.loc[df['Credit_History'].isnull(),'Credit_History']=df['Credit_History'].mean() 
```
