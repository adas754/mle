```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
df=pd.read_csv('loan_approved.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df['Self_Employed'] = df['Self_Employed'].fillna(df['Self_Employed'].mode()[0])
```

```python
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
df['Married'] = df['Married'].fillna(df['Married'].mode()[0])
```

```python
encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(
    df[['Gender', 'Married', 'Self_Employed']]
)

encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(),
    index=df.index
)
df = pd.concat([df, encoded_df], axis=1)
```

```python
df.head()
```

```python
#df_encoded=OrdinalEncoder(categories=[['Graduate', 'Not Graduate']])
encoder = OrdinalEncoder()
```

```python
df[['Education',
'Property_Area']] = encoder.fit_transform(df[['Education','Property_Area']])
```

```python
df['Education'].unique()
```

```python
df['Property_Area'].unique()
```

```python
encoder = LabelEncoder()
```

```python
df['Loan_Status (Approved)'] = encoder.fit_transform(df[['Loan_Status (Approved)']])
```

```python
df.drop(['Gender', 'Married', 'Self_Employed','Gender_Female',
         "Married_No","Self_Employed_No"], axis=1, inplace=True)
```
